# Practical 8: Interactive Visualizations Using Plotly

**Aim:** Create interactive visualizations using Plotly Express and Plotly Graph Objects.

## Step 1: Import Required Libraries

```python
# Import Pandas, Plotly Express, and Plotly Graph Objects
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

## Step 2: Create Dataset

```python
# Create sales records with updated values
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Product": ["Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop",
                "Mobile", "Tablet", "Laptop", "Mobile", "Tablet", "Laptop"],
    "Region": ["East", "West", "North", "South", "East", "West",
               "North", "South", "East", "West", "North", "South"],
    "Sales": [112000, 128000, 148000, 140000, 158000, 172000,
              185000, 168000, 195000, 215000, 225000, 248000],
    "Units_Sold": [18, 27, 23, 16, 32, 29, 21, 36, 34, 25, 43, 46],
    "Profit": [16000, 24000, 29000, 21000, 30000, 33000,
               27000, 31000, 36000, 34000, 42000, 48000]
}

# Convert sales records into a DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
```

## Step 4: Create an Interactive Line Chart Using Plotly Express

**Task:** Observe how sales revenue changes throughout the year.

```python
# Create an interactive line chart
fig = px.line(
    df,
    x='Month',
    y='Sales',
    title='Sales vs Month',
    markers=True
)

# Display the chart
fig.show()
```

## Step 5: Improve the Hover Information

**Task:** Show Product, Region, Units Sold, and Profit when hovering over a data point.

```python
# Create a line chart with additional hover information
fig = px.line(
    df,
    x='Month',
    y='Sales',
    title='Monthly Sales with Additional Information',
    markers=True,
    hover_data=['Product', 'Region', 'Units_Sold', 'Profit']
)

# Display the interactive chart
fig.show()
```

## Step 6: Create an Interactive Bar Chart

**Task:** Calculate and compare total sales generated by each product.

```python
# Group the dataset by Product and calculate total sales
grouped_by_product = df.groupby('Product')['Sales'].sum().reset_index()

# Create an interactive bar chart
fig = px.bar(
    grouped_by_product,
    x='Product',
    y='Sales',
    title='Total Sales by Product',
    labels={'Sales': 'Total Sales'}
)

# Display the chart
fig.show()
```

## Step 7: Create an Interactive Scatter Plot

**Task:** Analyze the relationship between units sold and sales.

```python
# Create an interactive scatter plot
fig = px.scatter(
    df,
    x='Units_Sold',
    y='Sales',
    title='Units Sold vs Sales',
    color='Product',
    hover_data=['Month', 'Region', 'Profit'],
    labels={'Units_Sold': 'Units Sold', 'Sales': 'Sales'}
)

# Display the chart
fig.show()
```

## Step 8: Create an Interactive Pie Chart

**Task:** Visualize the regional contribution to total sales.

```python
# Calculate total sales for each region
region_sales = df.groupby('Region', as_index=False)['Sales'].sum()

# Create an interactive pie chart
fig = px.pie(
    region_sales,
    values='Sales',
    names='Region',
    title='Sales by Region'
)

# Display the chart
fig.show()
```

## Step 9: Plotly Graph Objects

**Task:** Create a line chart using Plotly Graph Objects.

```python
# Create a Figure object
fig = go.Figure()

# Add a line trace using Scatter
fig.add_trace(
    go.Scatter(
        x=df['Month'],
        y=df['Sales'],
        mode='lines+markers',
        name='Sales'
    )
)

# Add chart title and axis labels
fig.update_layout(
    title='Sales (Plotly Graph Objects)',
    xaxis_title='Month',
    yaxis_title='Sales'
)

# Display the interactive chart
fig.show()
```
