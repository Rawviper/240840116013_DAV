# Practical 6: Create Basic Data Visualizations
**Aim:** Create basic data visualizations using Line, Bar, Histogram, Scatter, and Pie charts.

## Step 1: Import Libraries

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Suppress warnings
import warnings
warnings.filterwarnings('ignore')
```

## Step 2: Create a Dataset

```python
# Create monthly sales dataset with updated values
data = {
    "Month": ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
              'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    "Product": ['Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop',
                'Mobile', 'Tablet', 'Laptop', 'Mobile', 'Tablet', 'Laptop'],
    "Region": ['East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South'],
    "Sales": [110000, 128000, 142000, 138000, 155000, 168000,
              176000, 165000, 185000, 205000, 220000, 245000],
    "Units_Sold": [20, 28, 16, 24, 33, 27, 31, 36, 22, 39, 42, 48]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display dataset
df
```

## Step 3: Create a Line Chart
**Task:** Visualize company sales from January to December.

```python
# Set figure size
plt.figure(figsize=(10, 6))

# Plot monthly sales
plt.plot(df['Month'], df['Sales'], marker='o')

# Add grid, title, and labels
plt.grid(True, linestyle=':', linewidth=0.5)
plt.title('Monthly Company Sales')
plt.xlabel('Month')
plt.ylabel('Sales')

# Display chart
plt.show()
```

## Step 4: Create a Bar Chart
**Task:** Compare the total sales generated by each product.

```python
# Calculate total sales for each product
product_sales = df.groupby('Product')['Sales'].sum()

# Set figure size
plt.figure(figsize=(10, 6))

# Create bar chart
bars = plt.bar(product_sales.index, product_sales.values)

# Add title and labels
plt.title('Product-wise Total Sales')
plt.xlabel('Product')
plt.ylabel('Total Sales')

# Display total sales above each bar
plt.bar_label(bars, labels=[f'{value:,}' for value in product_sales.values])

# Display chart
plt.show()
```

## Step 5: Create a Histogram
**Task:** Observe the distribution of individual sales values.

```python
# Set figure size
plt.figure(figsize=(10, 6))

# Create histogram with KDE
sns.histplot(data=df, x='Sales', kde=True)

# Add title and labels
plt.title('Sales Distribution')
plt.xlabel('Sales')
plt.ylabel('Frequency')

# Display chart
plt.show()
```

## Step 6: Create a Scatter Plot
**Task:** Examine the relationship between units sold and total sales.

```python
# Set figure size
plt.figure(figsize=(10, 6))

# Create scatter plot
sns.scatterplot(data=df, x='Units_Sold', y='Sales')

# Add grid, title, and labels
plt.grid(True, linestyle=':', linewidth=0.5)
plt.title('Units Sold vs Sales')
plt.xlabel('Units Sold')
plt.ylabel('Sales')

# Display chart
plt.show()
```

## Step 7: Create a Pie Chart
**Task:** Show the percentage of total sales contributed by each region.

```python
# Calculate total sales for each region
region_sales = df.groupby('Region')['Sales'].sum()

# Set figure size
plt.figure(figsize=(10, 5))

# Create pie chart
plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct='%1.1f%%',
    startangle=140
)

# Add title and legend
plt.title('Region-wise Share of Total Sales')
plt.legend(title='Region')

# Display chart
plt.show()
```
