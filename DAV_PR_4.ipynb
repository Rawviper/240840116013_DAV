**Aim: Perform Exploratory Data Analysis (EDA) using descriptive statistics and identify trends in datasets.**

**Step 1: Import Libraries**

```python
# Import Pandas for data analysis
import pandas as pd

# Import Matplotlib for creating graphs
import matplotlib.pyplot as plt
```

**Step 2: Create Dataset**

```python
# Create a dictionary to store data
# Each key (Month, Units_Sold, Revenue, Profit) becomes a column
# Each list becomes the values of that column
data = {
    "Month": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Units_Sold": [120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Profit": [20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
# Convert dictionary into DataFrame
df = pd.DataFrame(data)

# Display dataset
df
```

**Step 3: Inspect the Dataset**

```python
# Display the first five rows
df.head()
```

```python
# Check number of rows and columns
df.shape
```

```python
# Display column names
df.columns
```

```python
# Display information about the DataFrame
df.info()
```

**Step 4: Generate Descriptive Statistics**

```python
# Generate descriptive statistics
df.describe()
```

**Step 5: Calculate the average number of units sold per month**

```python
# Calculate average number of units sold
average_units = df["Units_Sold"].mean()

print("Average units sold: ", average_units)
```

```python
# Select the Units_Sold column and calculate its median
median_units = df["Units_Sold"].median()

# Display the median units sold
print("Median of units sold: ", median_units)
```

**Step 7: Calculate max and min**

```python
# Import the minimum function from NumPy
from numpy import minimum

# Find the highest number of units sold
maximum_units = df["Units_Sold"].max()

# Find the lowest number of units sold
minimum_units = df["Units_Sold"].min()

# Display the maximum units sold
print("Maximum units sold: ", maximum_units)

# Display the minimum units sold
print("Minimum units sold: ", minimum_units)
```

**Step 8: Calculate Standard Deviation**

```python
# Calculate the standard deviation of units sold
std_units = df["Units_Sold"].std()

# Display the standard deviation of units sold
print("SD of units sold per month: ",std_units)
```

**Step 9: Find the month with higest Sales**

```python
#Find the index of the row having the max revenue
highest_sales_index = df['Revenue'].idxmax()

#Retrieve the month corresponding to that index
highest_sales_month = df.loc[highest_sales_index,"Month"]

# Retrieve the max revenue.
highest_revenue = df.loc[highest_sales_index,"Revenue"]

# Display the month with the highest sales
print("Month with highest sales: ", highest_sales_month)

# Display the highest revenue
print("Revenue: ", highest_revenue)
```

**Step 10 Find the month with Lowest Sales**

```python
# Find the index of the row with the lowest revenue
lowest_sales_index = df['Revenue'].idxmin()

# Get the month from the row with the lowest revenue
lowest_sales_month = df.loc[lowest_sales_index,"Month"]

# Get the revenue from the row with the lowest revenue
lowest_revenue = df.loc[lowest_sales_index,"Revenue"]

# Display the month with the lowest sales
print("Month with lowest sales: ", lowest_sales_month)

# Display the lowest revenue
print("Revenue: ", lowest_revenue)
```

**Step 11: Calculate Total Revenue**

```python
# Calculate the total revenue generated during the year
total_revenue = df["Revenue"].sum()

# Display the total revenue
print("Total Revenue: ", total_revenue)
```

**Step 12: Calculate Total Profit**

```python
# Calculate the total of all profit values
total_profit = df['Profit'].sum()

# Display the total profit
print("Total Profit: ", total_profit)
```

**Step 13: Identify the sales Trend**

```python
# Import lineStyles from Matplotlib
from matplotlib.lines import lineStyles

# Create a graph with size 10 by 5 inches
plt.figure(figsize=(10,5))

# Plot Month on X-axis and Revenue on Y-axis with circle markers and dashed line
plt.plot(df["Month"], df["Revenue"], marker='o', linestyle='--')

# Add a title to the graph
plt.title("Monthly Revenue Trend")

# Add the label Month to the X-axis
plt.xlabel("Month")

# Add the label Revenue to the Y-axis
plt.ylabel("Revenue")

# Rotate the month names by 45 degrees
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

**Step 14: Analyze Profit Trends**

```python
# Import lineStyles from Matplotlib
from matplotlib.lines import lineStyles

# Create a graph with width 10 and height 5
plt.figure(figsize=(10,5))

# Plot Month on X-axis and Profit on Y-axis
plt.plot(df["Month"], df["Profit"], marker='o', linestyle='--')

# Add a title to the graph
plt.title("Monthly Revenue Trend")

# Add a label to the X-axis
plt.xlabel("Month")

# Add a label to the Y-axis
plt.ylabel("Profit")

# Rotate the month names by 45 degrees
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

**Step 15: Compare Revenue and Profit**

```python
# Create a line chart to compare Revenue and Profit
# Create a figure with width 10 and height 5
plt.figure(figsize=(10,5))

# Plot monthly Revenue with circle markers and dashed line
plt.plot(df["Month"], df["Revenue"], marker='o', linestyle='--', label="Revenue")

# Plot monthly Profit with circle markers and dashed line
plt.plot(df["Month"], df["Profit"], marker='o', linestyle='--', label="Profit")

# Add a title to the comparison graph
plt.title("Comparison of Revenue and Profit")

# Add a label to the X-axis
plt.xlabel("Month")

# Add a label to the Y-axis
plt.ylabel("Amount")

# Display the labels of both lines
plt.legend()

# Rotate the month names by 45 degrees
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

**Step 16: Identify Growth in Sales**

```python
# Calculate the percentage change in Revenue from the previous month
df["Revenue_Growth_%"] = df['Revenue'].pct_change()*100

# Display the updated DataFrame
df
```

**Step 17: Identify Months with Increasing Revenue**

```python
# Select rows where revenue growth is greater than zero
increasing_sales = df[df['Revenue_Growth_%'] > 0]

# Display the months where sales increased
print(increasing_sales['Month'])
```
