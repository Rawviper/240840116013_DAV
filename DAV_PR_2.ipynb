**Aim: Perform data inspection, cleaning, handling missing values, duplicate removal, and data transformation using Pandas.**

**Step 1: Import Pandas**

```python
# Import the Pandas library
# Pandas is used to work with tables (DataFrames)
import pandas as pd
```

**Step 2:Create Sample Data**

```python
# Create a dictionary to store student data
# Each key (ID, Name, Age, etc.) becomes a column
# Each list becomes the values of that column
data = {
    "ID": [101,102,103,104,104,105,106,107],
    "Name": ["Rahul", "Priya", "Amit",None, "Karan", "Riya", "Aman","Ajay"],
    "Age": [20, 21, None, 20, 20, 22, 23, 28],
    "City": ["Surat", "Navsari", "Valsad", "Surat", "Gandhinagar", "Bharuch","Navsari","Ahemdabad"],
    "Marks": [85, None, 78, 88, 88, 81, 90, 91]
}
```

**Step 3: Create a Dataframe**

```python
# Convert the dictionary into a Pandas DataFrame
# A DataFrame represents data in rows and columns
df = pd.DataFrame(data)

# Display the DataFrame
df
```

**Step 4: Inspection**

Task 1: Display first five rows of dataset

```python
# Display the first 5 rows of the DataFrame
df.head()
```

Task 2: Display last five rows of dataset

```python
# Display the last 5 rows of the DataFrame
df.tail()
```

Task 3: Display detailed info about dataset

```python
# Display information about the DataFrame
# It shows columns, data types and non-empty values
df.info()
```

Task 4: Generate Statical info

```python
# Display statistical information about numerical columns
df.describe()
```

Task 5:  Display dimension of dataset

```python
# Display the number of rows and columns
df.shape
```

Task 6: Display the name of all columns

```python
# Display the names of all columns
df.columns
```

Task 7: Display the data type of each column

```python
# Display the data type of each column
df.dtypes
```

**Step 5: Check Missing Values**

```python
# Check for missing values in each column
df.isnull().sum()
```

**Step 6: Fill Missing Marks**

```python
# Replace missing Marks with the average Marks
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Display the updated DataFrame
df
```

**Step 7: Fill Missing Age**

```python
# Replace missing Age with the average Age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Display the updated DataFrame
df
```

**Step 8: Fill Missing Name**

```python
# Replace missing names with the word "Unknown"
df['Name'] = df['Name'].fillna("Unknown")

# Display the updated DataFrame
df
```

**Step 9: Detect Duplicat Records**

Task 1: check wheather each row is duplicated

```python
# Check which rows are duplicated
df.duplicated()
```

Task 2: Count total number of duplicate rows

```python
# Count the total number of duplicate rows
df.duplicated().sum()
```

**Step 10: Remove Duplicat Records**

```python
# Remove duplicate rows from the DataFrame
df = df.drop_duplicates()

# Display the DataFrame after removing duplicates
df
```

**Step 11: Rename Column**

```python
# Rename the "Marks" column to "Score"
df.rename(columns={"Marks": "Score"}, inplace=True)

# Display the updated DataFrame
df
```

**Step 12: Change the data type**

```python
# Convert Age column from float to integer
df['Age'] = df['Age'].astype(int)

# Display the updated DataFrame
df
```

**Step 13: Create a new column**

```python
# Create a new column named "Final Score"
# Add 5 marks to each student's Score
df['Final Score'] = df['Score'] + 5

# Display the updated DataFrame
df
```

**Step 14: Convert text into Uppercase**

```python
# Convert all Student names to uppercase
df['Name'] = df['Name'].str.upper()

# Display the updated DataFrame
df
```

**Step 15: Convert text to lowercase**

```python
# Convert all city names to lowercase
df['City'] = df['City'].str.lower()

# Display the updated DataFrame
df
```

**Step 16: Remove Extra Spaces**

```python
# Remove leading and trailing spaces from city names
df['City'] = df['City'].str.strip()

# Display the updated DataFrame
df
```

**Step 17: Sort the Dataset**

```python
# Sort the DataFrame using the Final Score column
# Sort from highest score to lowest score
df.sort_values(by='Final Score', ascending=False)
```

**Step 18: Save the cleaned Dataset**

```python
# Save the cleaned DataFrame as a new CSV file
# index=False prevents row numbers from being saved
df.to_csv("Cleaned_Students.csv",index=False)

# Display a success message
print("Dataset Saved Successfully.")
```
