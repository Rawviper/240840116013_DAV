## Step 9: Detect Outliers Using the IQR Method

```python
# Select the final_marks column
final_marks = df['final_marks']

# Calculate the first and third quartiles
Q1 = final_marks.quantile(0.25)
Q3 = final_marks.quantile(0.75)

# Calculate the interquartile range
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Display the calculated values
print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)

# Identify outlier records
outliers = df[
    (df['final_marks'] < lower_bound) |
    (df['final_marks'] > upper_bound)
]

# Display outlier student IDs and marks
print("\nOutliers:")
print(outliers[['student_id', 'final_marks']])
```

## Step 10: Identify and Display Outliers

### Task 10.1: Select records outside the IQR bounds

```python
# Select records below the lower bound or above the upper bound
outliers = df[
    (df['final_marks'] < lower_bound) |
    (df['final_marks'] > upper_bound)
]

# Display selected records
print("Records with final marks outside the bounds:")
print(outliers)
```

### Task 10.2: Display detected outlier records

```python
# Display the detected outlier records
print("Detected Outlier Records:")
print(outliers)
```

## Step 11: Count Outliers

```python
# Count the number of outliers
count_outliers = len(outliers)

# Display the count
print("Number of Outliers:", count_outliers)
```
